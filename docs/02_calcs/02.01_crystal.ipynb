## External Caps

```python
# Constants
CL = 12 * 10**(-12) # Load capacitance (as specified by JLC PCB)

# Stray capacitance (3-5pF)
CS_min = 3 * 10**(-12)
CS_max = 5 * 10**(-12)
```

```python
C_ext_min = 2 * (CL - CS_max)
C_ext_max = 2 * (CL - CS_min)
print(f"C_ext = [{C_ext_min}, {C_ext_max}]")
```
